# Случайные вектор и совместное распределение

```python
import numpy as np

import matplotlib.pyplot as plt

from IPython.display import display, Markdown
from tabulate import tabulate
```

```python
x = np.arange(0, 3)
y = np.arange(0, 4)

rng_prob = np.random.default_rng(2)
cum_probs = rng_prob.uniform(low=0, high=1, size=len(x)*len(y)-1).round(3)
cum_probs = np.append(arr=cum_probs, values=[0,1])
probs = np.reshape( np.diff(np.sort(cum_probs)), (len(x), len(y)) )

# print(probs)

# distr_law = np.array([x, probs])

display(Markdown('Рассмотрим случайный вектор с таблицей совместного распределения'))
display(Markdown(tabulate(probs, tablefmt='html', showindex=list(x), headers=list(y)) ))
```

Матрица вероятностей $\mathcal P = (f(x,y))$

```python
print(probs)
```

## Маргинальные распределения

Для $X$ маргинальные вероятности $f_1(x)=\sum_y f(x,y)$

```python
probs_X = probs.sum(axis=1)
probs_X
```

Математическое оживание $E[X]=\sum_x xf_1(x)$

```python
np.sum(x*probs_X)
```

Для $Y$ маргинальные вероятности $f_2(y)=\sum_x f(x,y)$

```python
probs_Y = probs.sum(axis=0)
probs_Y
```

Математическое оживание $E[Y]=\sum_y yf_2(y)$

```python
np.sum(y*probs_Y)
```

## Ковариация и корреляция

Вычислим $E[XY]=\sum_{x,y} xyf(x,y)=x^\top \mathcal{P} y$

```python
x.T@probs@y
```

Ковариация $cov[X,Y]=E[XY]-E[X]E[Y]$

```python
x.T@probs@y-np.sum(x*probs_X)*np.sum(y*probs_Y)
```

Корреляция 

$$
	corr[X,Y]=\frac{cov[X,Y]}{\sqrt{Var[X]Var[Y]}}
$$

```python
var_X = np.sum((x**2)*probs_X)-np.sum(x*probs_X)**2
var_Y = np.sum((y**2)*probs_Y)-np.sum(y*probs_Y)**2
cov_XY = x.T@probs@y-np.sum(x*probs_X)*np.sum(y*probs_Y)

cov_XY/((var_X*var_Y)**0.5)
```

## Условное распределение и условное математическое распределение

Найдем распределением $X$ при условии $Y=1$

$$
\begin{aligned}
	f_1(x|1)&=\frac{f(x,1)}{f_2(1)} & f_2(1)&=\sum_x f(x,1)
\end{aligned}
$$

```python
cond_probs_X = probs[:,y==1].T/np.sum(probs[:,y==1].T)
cond_probs_X
```

Условное математическое ожидание 

$$
	E(X|Y=1)=\sum_x xf_1(x|1)
$$

```python
np.sum(x*cond_probs_X)
```

Найдем распределением $Y$ при условии $X=2$

$$
\begin{aligned}
	f_2(y|2)&=\frac{f(2,y)}{f_1(2)} & f_1(2)&=\sum_y f(2,y)
\end{aligned}
$$

```python
cond_probs_Y = probs[x==2,:]/np.sum(probs[x==2,:])
cond_probs_Y
```

Условное математическое ожидание 

$$
	E(Y|X=2)=\sum_y yf_1(y|2)
$$

```python
np.sum(y*cond_probs_Y)
```
